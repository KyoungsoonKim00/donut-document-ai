```python
import os
from openpyxl import load_workbook
```

```python
# 1. PDF가 있는 폴더 지정
data_dir = r"data/raw_pdf"

# 2. 파일 리스트 가져오기 (pdf만)
pdf_files = [f for f in os.listdir(data_dir) if f.lower().endswith(".pdf")]

# 3. 저장할 엑셀 파일 경로
save_path = os.path.join(data_dir, "Labeling_Sample_Mokdong.xlsx")

# 4. 엑셀 파일이 존재하지 않으면 경고 후 종료
if not os.path.exists(save_path):
    print(f"❌ 엑셀 파일이 존재하지 않습니다: {save_path}")
    print("📌 먼저 엑셀 파일을 만들고 양식을 작성한 뒤 다시 실행해주세요.")
    exit()

# 5. 엑셀 파일 불러오기
wb = load_workbook(save_path)
ws = wb.active

# 6. A2셀부터 PDF 파일명 삽입
for idx, filename in enumerate(pdf_files, start=2):
    ws[f"A{idx}"] = filename

# 7. 저장
wb.save(save_path)

print(f"✅ PDF 파일명이 A2셀부터 삽입되었습니다: {save_path}")
```
